```python
%load_ext autoreload
%autoreload 2

import numpy as np
from simulator.simulator import Simulator
from model.thermal_network import ThermalNetwork
from control.hyst_control import HystControl
from control.hyst_21_control import Hyst21Control
from control.pid_control import PidControl
from control.pid_hyst_control import PidHystControl
```

```python
# simplistic model where instant heat transfer is assumed
model_simple = ThermalNetwork(T_init = 20.0)
model_simple.add_node("heater", C=1000.0)
model_simple.add_ambient_node("ambient", 20.0)
model_simple.add_link("heater", "ambient", R=3.0)
simulator_simple = Simulator(
    model=model_simple, 
    heater_power=50.0,
    heater_node="heater",
    sensor_node="heater",
)

# 2-node model where heat transfer from heater to air is simulated
model = ThermalNetwork(T_init = 20.0)
model.add_node("heater", C=250.0)
model.add_node("air", 150.0)
model.add_ambient_node("ambient", 20.0)
model.add_link("heater", "air", R=0.4)
model.add_link("air", "ambient", R=3.0)
simulator = Simulator(
    model=model, 
    heater_power=50.0,
    heater_node="heater",
    sensor_node="air",
)
```

```python

control = HystControl(
    hyst_low=-0.5, 
    hyst_high=0.5, 
    output_low=1.0,
    output_high=0.0
)
simulator_simple.simulate(
    control=control,
    samples=2000,
    dt=1,
)
```
