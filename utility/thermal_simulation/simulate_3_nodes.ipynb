```python
%load_ext autoreload
%autoreload 2

import numpy as np
from simulator.simulator import Simulator
from model.thermal_network import ThermalNetwork
from control.hyst_control import HystControl
from control.pid_control import PidControl
```

```python
# 2-node model where heat transfer from heater to air is simulated
model = ThermalNetwork(T_init = 20.0)

# add nodes
model.add_node("Heater", C=200.0)
model.add_node("Air",    C=120.0)
model.add_node("Dough",  C=4000.0)

# add ambient node
model.add_ambient_node("Ambient", 20.0)

# add links
model.add_link("Heater", "Air",     R=0.5)
model.add_link("Air",    "Ambient", R=4.0)
model.add_link("Air",    "Dough",   R=2.0)

# create simulator
simulator = Simulator(
    model=model, 
    heater_power=50.0,
    heater_node="Heater",
    sensor_node="Air",
)
```

```python
control = HystControl(
    hyst_low=-0.5, 
    hyst_high=0.5, 
    output_low=1.0,
    output_high=0.0
)
simulator.simulate(
    control=control,
    samples=2000,
    dt=1,
)
```

```python
control = PidControl(Kp=0.25, Ki=0.002, Kd=0)
simulator.simulate(
    control=control,
    samples=2000,
    dt=1,
    plot_slicing=1
)
```

```python
samples = 5000
temp_target_list = np.ones(samples, dtype=np.float64) * 30.0
temp_target_list[1200:2400] = 40.0
temp_target_list[3600:] = 25.0
control = PidControl(Kp=0.25, Ki=0.002, Kd=0)
simulator.simulate(
    control=control,
    samples=samples,
    dt=1,
    temp_target_list=temp_target_list
)
```
